The dataset originates from the **Kaggle NeuroBehavior Clinical Health Risk Sample Dataset** and contains 1,000 patient records with 23 clinical and behavioral features spanning multiple health systems:

- **Behavioral Indicators** (7 features): Sleep duration, stress intensity, screen time exposure, physical activity, meditation practice, social engagement, rumination patterns

- **Demographics** (2 features): Age, gender

# SECTION 5: Clinical Domain Context & System-Level Understanding
---

## Human Health as a Complex Adaptive System

Human health is governed by interconnected biological systems. This dataset captures key markers from:

### System Relationships
- **Stress --> Cortisol --> Inflammation**: Chronic stress elevates cortisol, triggering systemic inflammation (CRP, ESR)
- **Sleep --> Neural Restoration**: Adequate sleep restores cognitive function (Focus_Level, Mood_Score)
- **Metabolic Health**: Fasting glucose, insulin, and lipid profiles indicate metabolic syndrome risk
- **Neurochemical Balance**: Dopamine and oxytocin levels reflect reward system and social bonding capacity
- **Liver Function**: ALT/AST elevation indicates hepatic stress from metabolic overload

### Clinical Insight
Health risk is a **complex adaptive system outcome**, not a linear function of any single variable.


````python
# Visualization 2: Clinical Systems Pathway Diagram
fig, ax = plt.subplots(1, 1, figsize=(14, 8))
ax.set_xlim(0, 10)
ax.set_ylim(0, 10)
ax.axis('off')
ax.set_title('Clinical Systems Pathway Diagram', fontsize=16, fontweight='bold', pad=20)

# System boxes
systems = {
    'Stress System': (1.5, 8, '#e74c3c'),
    'Sleep System': (5, 8, '#3498db'),
    'Metabolic System': (8.5, 8, '#f39c12'),
    'Inflammation': (3, 5, '#e67e22'),
    'Neurochemical': (7, 5, '#9b59b6'),
    'Liver Function': (5, 2, '#1abc9c'),
    'HEALTH RISK': (5, 0.5, '#2c3e50')
}

for name, (x, y, color) in systems.items():
    box = mpatches.FancyBboxPatch((x - 1, y - 0.4), 2, 0.8, boxstyle="round,pad=0.1",
                                   facecolor=color, alpha=0.85, edgecolor='white', linewidth=2)
    ax.add_patch(box)
    fontsize = 11 if name != 'HEALTH RISK' else 13
    ax.text(x, y, name, ha='center', va='center', fontsize=fontsize,
            fontweight='bold', color='white')

# Arrows
arrow_pairs = [
    ((1.5, 7.5), (3, 5.5)),   # Stress -> Inflammation
    ((5, 7.5), (7, 5.5)),     # Sleep -> Neurochemical
    ((5, 7.5), (3, 5.5)),     # Sleep -> Inflammation
    ((8.5, 7.5), (7, 5.5)),   # Metabolic -> Neurochemical
    ((8.5, 7.5), (5, 2.5)),   # Metabolic -> Liver
    ((3, 4.5), (5, 1)),       # Inflammation -> Risk
    ((7, 4.5), (5, 1)),       # Neurochemical -> Risk
    ((5, 1.6), (5, 1)),       # Liver -> Risk
]
for (x1, y1), (x2, y2) in arrow_pairs:
    ax.annotate('', xy=(x2, y2), xytext=(x1, y1),
                arrowprops=dict(arrowstyle='->', color='#7f8c8d', lw=2))

plt.tight_layout()
plt.savefig('viz_02_clinical_pathway.png', dpi=150, bbox_inches='tight')
plt.show()
print("[Visualization 2] Clinical Systems Pathway Diagram")
````

````python
# SECTION 9: Data Preprocessing
print("DATA PREPROCESSING PIPELINE")
print("=" * 60)

# 1. Remove ID column for analysis
df_clean = df.drop('ID', axis=1).copy()
print("1. Dropped 'ID' column (non-analytical)")

# 2. Encode Gender
le_gender = LabelEncoder()
df_clean['Gender_Encoded'] = le_gender.fit_transform(df_clean['Gender'])
print(f"2. Encoded Gender: {dict(zip(le_gender.classes_, le_gender.transform(le_gender.classes_)))}")

# 3. Handle any missing values
if df_clean.isnull().sum().sum() > 0:
    for col in df_clean.select_dtypes(include=[np.number]).columns:
        df_clean[col].fillna(df_clean[col].median(), inplace=True)
    print("3. Missing values imputed with median")
else:
    print("3. No missing values to handle")

# 4. Remove exact duplicates
n_before = len(df_clean)
df_clean = df_clean.drop_duplicates()
print(f"4. Duplicates removed: {n_before - len(df_clean)} records")

# 5. Create Risk Label from clinical biomarkers
# High CRP (>3), High Fasting Glucose (>100), High LDL (>130), Low HDL (<40),
# High Triglycerides (>200), High ALT (>56)
df_clean['Risk_Score'] = 0
df_clean.loc[df_clean['CRP'] > 3, 'Risk_Score'] += 1
df_clean.loc[df_clean['Fasting_Glucose'] > 110, 'Risk_Score'] += 1
df_clean.loc[df_clean['LDL'] > 150, 'Risk_Score'] += 1
df_clean.loc[df_clean['HDL'] < 40, 'Risk_Score'] += 1
df_clean.loc[df_clean['Triglycerides'] > 200, 'Risk_Score'] += 1
df_clean.loc[df_clean['ALT'] > 55, 'Risk_Score'] += 1
df_clean.loc[df_clean['Stress_Level'] >= 7, 'Risk_Score'] += 1
df_clean.loc[df_clean['Sleep_Hours'] < 5, 'Risk_Score'] += 1

df_clean['Risk_Label'] = pd.cut(df_clean['Risk_Score'], bins=[-1, 2, 4, 10],
                                 labels=['Low', 'Medium', 'High'])
print(f"5. Engineered Risk_Label from clinical biomarkers")
print(f"   Distribution: {df_clean['Risk_Label'].value_counts().to_dict()}")

print(f"\nFinal cleaned dataset shape: {df_clean.shape}")
display(df_clean.head())
````

````python
# SECTION 17: Feature Engineering
print("FEATURE ENGINEERING PIPELINE")
print("=" * 60)

# 1. Stress-Sleep Ratio
df_clean['Stress_Sleep_Ratio'] = df_clean['Stress_Level'] / (df_clean['Sleep_Hours'] + 0.1)
print("1. Created Stress_Sleep_Ratio = Stress_Level / Sleep_Hours")

# 2. LDL/HDL Ratio (Atherogenic Index)
df_clean['LDL_HDL_Ratio'] = df_clean['LDL'] / (df_clean['HDL'] + 0.1)
print("2. Created LDL_HDL_Ratio (Atherogenic Index)")

# 3. ALT/AST Ratio (De Ritis Ratio)
df_clean['DeRitis_Ratio'] = df_clean['AST'] / (df_clean['ALT'] + 0.1)
print("3. Created DeRitis_Ratio (AST/ALT)")

# 4. Triglyceride-Glucose Index (TyG)
df_clean['TyG_Index'] = np.log(df_clean['Triglycerides'] * df_clean['Fasting_Glucose'] / 2)
print("4. Created TyG_Index (Triglyceride-Glucose Index)")

# 5. Composite Health Index
df_clean['Health_Index'] = (
    (df_clean['Sleep_Hours'] / 9.0) * 20 +
    ((10 - df_clean['Stress_Level']) / 10.0) * 20 +
    (df_clean['Physical_Activity'] / 8.0) * 15 +
    ((10 - df_clean['CRP'] / df_clean['CRP'].max() * 10)) * 15 +
    (df_clean['HDL'] / df_clean['HDL'].max()) * 15 +
    (df_clean['Mood_Score'] / 10.0) * 15
)
print("5. Created Health_Index (Composite 0-100)")

# 6. Behavioral Wellness Score
df_clean['Behavioral_Wellness'] = (
    df_clean['Sleep_Hours'] * 2 +
    (10 - df_clean['Stress_Level']) * 1.5 +
    df_clean['Physical_Activity'] * 1.5 +
    df_clean['Meditation_Minutes'] / 10 +
    df_clean['Social_Interaction'] * 1
)
print("6. Created Behavioral_Wellness score")

# 7. Inflammatory Burden
df_clean['Inflammatory_Burden'] = (
    df_clean['CRP'] / df_clean['CRP'].max() * 50 +
    df_clean['ESR'] / df_clean['ESR'].max() * 50
)
print("7. Created Inflammatory_Burden (normalized 0-100)")

# 8. Metabolic Risk Score
df_clean['Metabolic_Risk'] = (
    (df_clean['Fasting_Glucose'] > 100).astype(int) +
    (df_clean['Insulin'] > 15).astype(int) +
    (df_clean['LDL'] > 130).astype(int) +
    (df_clean['HDL'] < 40).astype(int) +
    (df_clean['Triglycerides'] > 150).astype(int)
)
print("8. Created Metabolic_Risk (count of metabolic syndrome markers)")

engineered_features = ['Stress_Sleep_Ratio', 'LDL_HDL_Ratio', 'DeRitis_Ratio', 'TyG_Index',
                       'Health_Index', 'Behavioral_Wellness', 'Inflammatory_Burden', 'Metabolic_Risk']
print(f"\nTotal new features: {len(engineered_features)}")
print(f"Updated dataset shape: {df_clean.shape}")
display(df_clean[engineered_features].describe().round(2))
````

````python
# SECTION 18: Statistical Tests
print("STATISTICAL HYPOTHESIS TESTING")
print("=" * 60)

# Separate by gender
male_data = df_clean[df_clean['Gender'] == 'Male']
female_data = df_clean[df_clean['Gender'] == 'Female']

# T-tests: Gender differences
print("\n--- Independent Samples t-tests (Male vs Female) ---")
ttest_results = {}
test_features = ['Stress_Level', 'Sleep_Hours', 'CRP', 'Fasting_Glucose', 'LDL', 'HDL', 'Triglycerides']
for feat in test_features:
    t_stat, p_val = ttest_ind(male_data[feat].dropna(), female_data[feat].dropna())
    sig = "***" if p_val < 0.001 else "**" if p_val < 0.01 else "*" if p_val < 0.05 else "ns"
    ttest_results[feat] = {'t-stat': t_stat, 'p-value': p_val, 'significance': sig}
    print(f"  {feat:20s}: t={t_stat:>7.3f}, p={p_val:.4f} [{sig}]")

# ANOVA: Risk Label groups
print("\n--- One-Way ANOVA (Risk Groups: Low vs Medium vs High) ---")
anova_results = {}
groups = [df_clean[df_clean['Risk_Label'] == lbl] for lbl in ['Low', 'Medium', 'High']]
for feat in test_features:
    f_stat, p_val = f_oneway(*[g[feat].dropna() for g in groups])
    sig = "***" if p_val < 0.001 else "**" if p_val < 0.01 else "*" if p_val < 0.05 else "ns"
    anova_results[feat] = {'F-stat': f_stat, 'p-value': p_val, 'significance': sig}
    print(f"  {feat:20s}: F={f_stat:>8.3f}, p={p_val:.4f} [{sig}]")

# Shapiro-Wilk normality test
print("\n--- Shapiro-Wilk Normality Test (sample of 500) ---")
for feat in ['Sleep_Hours', 'CRP', 'Fasting_Glucose']:
    sample = df_clean[feat].dropna().sample(min(500, len(df_clean)), random_state=42)
    w_stat, p_val = shapiro(sample)
    normal = "Normal" if p_val > 0.05 else "Non-Normal"
    print(f"  {feat:20s}: W={w_stat:.4f}, p={p_val:.4f} --> {normal}")

# Mann-Whitney U test
print("\n--- Mann-Whitney U Test (Male vs Female) ---")
for feat in ['Stress_Level', 'CRP', 'Sleep_Hours']:
    u_stat, p_val = mannwhitneyu(male_data[feat].dropna(), female_data[feat].dropna())
    sig = "Significant" if p_val < 0.05 else "Not Significant"
    print(f"  {feat:20s}: U={u_stat:>10.1f}, p={p_val:.4f} --> {sig}")

# Chi-square test
print("\n--- Chi-Square Test (Gender vs Risk Label) ---")
ct = pd.crosstab(df_clean['Gender'], df_clean['Risk_Label'])
chi2, p_val, dof, expected = chi2_contingency(ct)
print(f"  Chi2={chi2:.3f}, p={p_val:.4f}, dof={dof}")
print(f"  Association: {'Significant' if p_val < 0.05 else 'Not Significant'}")
````

````python
# SECTION 19: Outlier Detection
print("OUTLIER DETECTION ANALYSIS")
print("=" * 60)

outlier_results = {}
for col in all_numeric:
    q1 = df_clean[col].quantile(0.25)
    q3 = df_clean[col].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    outliers = ((df_clean[col] < lower) | (df_clean[col] > upper)).sum()
    z_outliers = (np.abs(stats.zscore(df_clean[col].dropna())) > 3).sum()
    outlier_results[col] = {'IQR_Outliers': outliers, 'Zscore_Outliers': z_outliers,
                            'IQR_Pct': round(outliers/len(df_clean)*100, 2)}
    print(f"  {col:25s}: IQR={outliers:>3d} ({outlier_results[col]['IQR_Pct']:.1f}%), Z-score={z_outliers:>3d}")
````

````python
# SECTION 32: K-Means Clustering
print("CLUSTERING ANALYSIS")
print("=" * 60)

cluster_features = ['Sleep_Hours', 'Stress_Level', 'CRP', 'Fasting_Glucose',
                     'LDL', 'HDL', 'Triglycerides']
X_cluster = df_clean[cluster_features].copy().fillna(0)
scaler_cl = StandardScaler()
X_cluster_scaled = scaler_cl.fit_transform(X_cluster)

# Elbow method
inertias = []
sil_scores = []
K_range = range(2, 11)
for k in K_range:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    km.fit(X_cluster_scaled)
    inertias.append(km.inertia_)
    sil_scores.append(silhouette_score(X_cluster_scaled, km.labels_))

optimal_k = list(K_range)[np.argmax(sil_scores)]
print(f"Optimal K (max silhouette): {optimal_k}")

# Final clustering
km_final = KMeans(n_clusters=optimal_k, random_state=42, n_init=10)
df_clean['Cluster'] = km_final.fit_predict(X_cluster_scaled)
print(f"Cluster distribution:\n{df_clean['Cluster'].value_counts().sort_index()}")
````

````python
# Visualization 78-79: Elbow & Silhouette
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

axes[0].plot(list(K_range), inertias, 'bo-', linewidth=2, markersize=8)
axes[0].set_xlabel('Number of Clusters (K)', fontsize=12)
axes[0].set_ylabel('Inertia (Within-Cluster SS)', fontsize=12)
axes[0].set_title('Elbow Method -- Optimal K Selection', fontsize=14, fontweight='bold')
axes[0].grid(True, alpha=0.3)

axes[1].plot(list(K_range), sil_scores, 'rs-', linewidth=2, markersize=8)
axes[1].axvline(x=optimal_k, color='#2ecc71', linestyle='--', linewidth=2, label=f'Optimal K={optimal_k}')
axes[1].set_xlabel('Number of Clusters (K)', fontsize=12)
axes[1].set_ylabel('Silhouette Score', fontsize=12)
axes[1].set_title('Silhouette Score vs K', fontsize=14, fontweight='bold')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('viz_78_79_elbow_silhouette.png', dpi=150, bbox_inches='tight')
plt.show()
print("[Visualization 78-79] Elbow & Silhouette Analysis")
````

````python
# Visualization 85: Feature Interaction Network (Correlation-based)
fig, ax = plt.subplots(figsize=(12, 12))
import matplotlib.lines as mlines

network_features = ['Sleep_Hours', 'Stress_Level', 'CRP', 'Fasting_Glucose',
                     'LDL', 'HDL', 'Mood_Score', 'Physical_Activity']
n = len(network_features)
angles = np.linspace(0, 2*np.pi, n, endpoint=False)
positions = {f: (3*np.cos(a), 3*np.sin(a)) for f, a in zip(network_features, angles)}

# Draw edges (correlations)
corr_net = df_clean[network_features].corr()
for i, f1 in enumerate(network_features):
    for j, f2 in enumerate(network_features):
        if i < j and abs(corr_net.loc[f1, f2]) > 0.05:
            x1, y1 = positions[f1]
            x2, y2 = positions[f2]
            color = '#e74c3c' if corr_net.loc[f1, f2] < 0 else '#2ecc71'
            width = abs(corr_net.loc[f1, f2]) * 3
            ax.plot([x1, x2], [y1, y2], color=color, linewidth=width, alpha=0.4)

# Draw nodes
for f, (x, y) in positions.items():
    circle = plt.Circle((x, y), 0.5, color='#3498db', alpha=0.9, ec='white', lw=2)
    ax.add_patch(circle)
    ax.text(x, y, f.replace('_', '\n'), ha='center', va='center', fontsize=7,
            fontweight='bold', color='white')

ax.set_xlim(-5, 5)
ax.set_ylim(-5, 5)
ax.set_aspect('equal')
ax.axis('off')
ax.set_title('Feature Interaction Network\n(Green=Positive, Red=Negative correlation)',
             fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('viz_85_interaction_network.png', dpi=150, bbox_inches='tight')
plt.show()
print("[Visualization 85] Feature Interaction Network")
````
